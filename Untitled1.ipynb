```python

import numpy as np
import matplotlib.pyplot as plt
import time
from PIL import Image



IMAGE_PATH = None


def load_image():
    if IMAGE_PATH:
        img = Image.open(IMAGE_PATH).convert("L")
        img = img.resize((256, 256))
        return np.array(img)


    np.random.seed(42)
    img = np.zeros((256, 256))

    img[:128, :128] = np.random.normal(50, 10, (128, 128))
    img[:128, 128:] = np.random.normal(110, 12, (128, 128))
    img[128:, :128] = np.random.normal(170, 12, (128, 128))
    img[128:, 128:] = np.random.normal(220, 10, (128, 128))

    return np.clip(img, 0, 255).astype(np.uint8)


image = load_image()



histogram, _ = np.histogram(
    image.flatten(), bins=256, range=(0, 256)
)

prob = histogram / histogram.sum()
gray_levels = np.arange(256)

def fitness(chromosome):
    thresholds = np.sort(
        np.clip(np.round(chromosome).astype(int), 1, 254)
    )

    boundaries = [0] + list(thresholds) + [256]

    total_variance = 0

    for i in range(len(boundaries) - 1):
        start = boundaries[i]
        end = boundaries[i + 1]

        class_prob = prob[start:end]
        weight = class_prob.sum()

        if weight == 0:
            continue

        mean = np.sum(
            gray_levels[start:end] * class_prob
        ) / weight

        total_variance += weight * (
            mean - np.sum(gray_levels * prob)
        ) ** 2

    return total_variance




POPULATION_SIZE = 50
GENERATIONS = 100
NUM_THRESHOLDS = 2
MUTATION_RATE = 0.15
CROSSOVER_RATE = 0.85

def create_population():
    return np.sort(
        np.random.randint(
            1, 255,
            (POPULATION_SIZE, NUM_THRESHOLDS)
        ),
        axis=1
    )


def tournament_selection(population, scores):
    selected = []

    for _ in range(2):
        candidates = np.random.choice(
            len(population), 3, replace=False
        )
        winner = candidates[np.argmax(scores[candidates])]
        selected.append(population[winner])

    return selected


def rank_selection(population, scores):

    order = np.argsort(scores)
    ranks = np.empty(len(scores))
    ranks[order] = np.arange(1, len(scores) + 1)

    probabilities = ranks / ranks.sum()

    indices = np.random.choice(
        len(population),
        size=2,
        replace=True,
        p=probabilities
    )

    return population[indices[0]], population[indices[1]]


def crossover(parent1, parent2):
    alpha = np.random.random()

    child1 = alpha * parent1 + (1 - alpha) * parent2
    child2 = alpha * parent2 + (1 - alpha) * parent1

    return child1, child2


def mutate(chromosome):
    for i in range(NUM_THRESHOLDS):
        if np.random.random() < MUTATION_RATE:
            chromosome[i] += np.random.normal(0, 15)

    return np.sort(np.clip(chromosome, 1, 254))




def run_ga(improved=True, generations=GENERATIONS, seed=42):

    np.random.seed(seed)

    population = create_population()

    best_history = []

    start_time = time.time()

    for generation in range(generations):

        scores = np.array([
            fitness(individual)
            for individual in population
        ])

        best_index = np.argmax(scores)

        best_history.append(scores[best_index])

        new_population = []


        if improved:
            elite_indices = np.argsort(scores)[-2:]
            for index in elite_indices:
                new_population.append(
                    population[index].copy()
                )

        while len(new_population) < POPULATION_SIZE:

            if improved:
                p1, p2 = rank_selection(population, scores)
            else:
                p1, p2 = tournament_selection(population, scores)

            if np.random.random() < CROSSOVER_RATE:
                child1, child2 = crossover(p1, p2)
            else:
                child1, child2 = p1.copy(), p2.copy()

            child1 = mutate(child1)
            child2 = mutate(child2)

            new_population.append(child1)

            if len(new_population) < POPULATION_SIZE:
                new_population.append(child2)

        population = np.array(new_population)

    final_scores = np.array([
        fitness(individual)
        for individual in population
    ])

    best_index = np.argmax(final_scores)

    best_thresholds = np.sort(
        np.round(population[best_index]).astype(int)
    )

    elapsed_time = time.time() - start_time

    return (
        best_thresholds,
        best_history,
        elapsed_time
    )




basic_thresholds, basic_history, basic_time = run_ga(
    improved=False
)

improved_thresholds, improved_history, improved_time = run_ga(
    improved=True
)

print("Basic GA Thresholds:", basic_thresholds)
print("Improved GA Thresholds:", improved_thresholds)

print("Basic GA Time:", round(basic_time, 4), "seconds")
print("Improved GA Time:", round(improved_time, 4), "seconds")

print("Basic GA Fitness:", round(max(basic_history), 4))
print("Improved GA Fitness:", round(max(improved_history), 4))




def segment_image(img, thresholds):
    segmented = np.zeros_like(img)

    t1, t2 = thresholds

    segmented[img <= t1] = 50
    segmented[(img > t1) & (img <= t2)] = 125
    segmented[img > t2] = 220

    return segmented


segmented_image = segment_image(
    image, improved_thresholds
)




plt.figure(figsize=(8, 5))

plt.plot(basic_history, label="Basic GA")
plt.plot(improved_history, label="Improved GA")

plt.xlabel("Generations")
plt.ylabel("Best Fitness")
plt.title("GA Convergence Comparison")

plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig("graph1_convergence.png")
plt.show()




plt.figure(figsize=(7, 5))

methods = ["Basic GA", "Improved GA"]

values = [
    max(basic_history),
    max(improved_history)
]

plt.bar(methods, values)

plt.ylabel("Best Fitness")
plt.title("Final Fitness Comparison")

plt.tight_layout()
plt.savefig("graph2_fitness_comparison.png")
plt.show()




runs = 5

all_runs = []

for seed in range(runs):

    _, history, _ = run_ga(
        improved=True,
        seed=seed
    )

    all_runs.append(history)

plt.figure(figsize=(8, 5))

for i, history in enumerate(all_runs):
    plt.plot(history, label=f"Run {i + 1}")

plt.xlabel("Generations")
plt.ylabel("Best Fitness")
plt.title("GA Stability Across Multiple Runs")

plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig("graph3_stability.png")
plt.show()




generation_values = [20, 40, 60, 80, 100]
execution_times = []

for gen in generation_values:

    _, _, elapsed = run_ga(
        improved=True,
        generations=gen,
        seed=42
    )

    execution_times.append(elapsed)

plt.figure(figsize=(8, 5))

plt.plot(
    generation_values,
    execution_times,
    marker="o"
)

plt.xlabel("Number of Generations")
plt.ylabel("Execution Time (Seconds)")
plt.title("Running Time vs Generations")

plt.grid(True)
plt.tight_layout()
plt.savefig("graph4_running_time.png")
plt.show()



plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(segmented_image, cmap="gray")
plt.title("GA Segmented Image")
plt.axis("off")

plt.tight_layout()
plt.savefig("graph5_segmentation.png")
plt.show()

print("\nAll graphs generated successfully!")
```
