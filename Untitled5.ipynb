```python
import matplotlib.pyplot as plt
import numpy as np

automata = ["GoL", "Maze", "Brian", "Fire", "Wire", "Excitable", "Cyclic", "Fluid", "Critters", "Traffic", "Geo Mean"]

bit_planes_linear = [4.9, 5.0, 3.2, 3.5, 3.1, 2.3, 1.2, 1.8, 3.1, 3.5, 3.0]
bit_planes_tiled = [4.2, 4.0, 2.8, 3.2, 2.7, 2.0, 1.1, 1.8, 2.5, 3.4, 2.8]
temporal_linear = [6.6, 6.9, 4.3, 7.4, 4.1, 3.2, 1.4, 5.6, 4.7, 10.8, 4.9]
temporal_tiled = [5.4, 5.6, 3.1, 7.1, 3.3, 2.2, 1.5, 9.2, 2.5, 12.2, 4.2]

x = np.arange(len(automata))
width = 0.19

plt.figure(figsize=(13, 6))

plt.bar(x - 1.5 * width, bit_planes_linear, width, label="Bit Planes (Linear)")
plt.bar(x - 0.5 * width, bit_planes_tiled, width, label="Bit Planes (Tiled)")
plt.bar(x + 0.5 * width, temporal_linear, width, label="Temporal (Linear)")
plt.bar(x + 1.5 * width, temporal_tiled, width, label="Temporal (Tiled)")

plt.xlabel("Cellular Automata")
plt.ylabel("Throughput (10^12 cells/sec)")
plt.title("Throughput Comparison of Linear and Tiled Bit-Plane Layouts")

plt.xticks(x, automata)
plt.grid(axis="y", linestyle=":")
plt.legend()
plt.tight_layout()

plt.show()
```
